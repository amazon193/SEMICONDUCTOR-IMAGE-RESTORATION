```python
import sys
sys.path.append('/content/semiconductor-image-restoration/src')  # after cloning repo, or adjust path
# If running standalone in Colab without cloning the repo, copy src/preprocessing.py contents here instead.
from preprocessing import verify_dataset

root_dir = "/content/dataset/train"  # adjust to actual extracted path
verify_dataset(root_dir)
```

```python
import numpy as np, os
noisy_dir = os.path.join(root_dir, "NoisyLR")
clean_dir = os.path.join(root_dir, "CleanHR")

sample = np.load(os.path.join(noisy_dir, sorted(os.listdir(noisy_dir))[0]))
print("shape:", sample.shape, "dtype:", sample.dtype, "min/max:", sample.min(), sample.max())
```

```python
import matplotlib.pyplot as plt

files = sorted(os.listdir(noisy_dir))[:4]
fig, axes = plt.subplots(2, 4, figsize=(14, 7))
for i, f in enumerate(files):
    noisy = np.load(os.path.join(noisy_dir, f))
    clean = np.load(os.path.join(clean_dir, f))
    axes[0, i].imshow(noisy, cmap='gray'); axes[0, i].set_title(f"Degraded {f}"); axes[0, i].axis('off')
    axes[1, i].imshow(clean, cmap='gray'); axes[1, i].set_title(f"Clean {f}"); axes[1, i].axis('off')
plt.tight_layout()
plt.show()
```
